```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix
)
from catboost import CatBoostClassifier
```

```python
RANDOM_STATE = 42
REVIEW_BUDGET = 0.15
```

```python
users=pd.read_csv('users.csv')
orders=pd.read_csv('orders.csv')
claims_train=pd.read_csv('claims_train.csv')
claims_test=pd.read_csv('claims_test.csv')
labels=pd.read_csv('train_labels.csv')
history=pd.read_csv('claim_history_outcomes.csv')
print(f'users: {users.shape}')
print(f'orders: {orders.shape}')
print(f'claims_train: {claims_train.shape}')
print(f'claims_test: {claims_test.shape}')
print(f'labels: {labels.shape}')
print(f'history: {history.shape}')
```

```python
df = df.merge(users[['user_id','registered_at','country','device_type','is_verified']],
              on='user_id', how='left')

df['created_at'] = pd.to_datetime(df['created_at'])
df['registered_at'] = pd.to_datetime(df['registered_at'])

df['account_age_days'] = (df['created_at'] - df['registered_at']).dt.days

df = df.merge(
    orders[['order_id','created_at','delivered_at','product_category',
            'order_amount','payment_method','delivery_days']]
    .rename(columns={'created_at':'order_created_at','delivered_at':'order_delivered_at'}),
    on='order_id', how='left')
df['order_created_at'] = pd.to_datetime(df['order_created_at'])
df['order_delivered_at'] = pd.to_datetime(df['order_delivered_at'])

df['claim_to_order_ratio']        = df['claim_amount'] / df['order_amount'].replace(0, np.nan)
df['days_from_order_to_claim']    = (df['created_at'] - df['order_created_at']).dt.days
df['days_from_delivery_to_claim'] = (df['created_at'] - df['order_delivered_at']).dt.days
```

```python
orders_s = orders.sort_values(['user_id','created_at']).copy()
orders_s['orders_cum'] = orders_s.groupby('user_id').cumcount() + 1
orders_s['amount_cum'] = orders_s.groupby('user_id')['order_amount'].cumsum()
orders_s['amount_max'] = orders_s.groupby('user_id')['order_amount'].cummax()
orders_s['created_at'] = pd.to_datetime(orders_s['created_at'])
df['created_at'] = pd.to_datetime(df['created_at'])

orders_s = orders_s.sort_values('created_at')

m = pd.merge_asof(
    df[['claim_id','user_id','created_at']].sort_values('created_at'),
    orders_s[['user_id','created_at','orders_cum','amount_cum','amount_max']],
    on='created_at', by='user_id', direction='backward', allow_exact_matches=False
).set_index('claim_id')

df['total_past_orders']  = df['claim_id'].map(m['orders_cum']).fillna(0).astype(int)
df['total_order_amount'] = df['claim_id'].map(m['amount_cum']).fillna(0)
df['max_order_amount']   = df['claim_id'].map(m['amount_max'])
df['avg_order_amount']   = df['total_order_amount'] / df['total_past_orders'].replace(0, np.nan)

m_last = pd.merge_asof(
    df[['claim_id','user_id','created_at']].sort_values('created_at'),
    orders_s[['user_id','created_at']].rename(columns={'created_at':'last_order_ts'}),
    left_on='created_at', right_on='last_order_ts', 
    by='user_id', direction='backward', allow_exact_matches=False
).set_index('claim_id')

df['days_since_last_order'] = (df['created_at'] - df['claim_id'].map(m_last['last_order_ts'])) / np.timedelta64(1, 'D')
df['orders_per_active_month'] = df['total_past_orders'] / (df['account_age_days'] / 30).clip(lower=1)
```

```python
history_full = history.merge(df[['claim_id', 'user_id']], on='claim_id', how='left')

all_claims = pd.concat([
    df[['user_id', 'created_at']], 
    history_full[['user_id', 'decision_at']].rename(columns={'decision_at': 'created_at'})
]).drop_duplicates().sort_values('created_at').reset_index(drop=True)

all_claims['claims_cum'] = all_claims.groupby('user_id').cumcount() + 1

m_c = pd.merge_asof(
    df[['claim_id','user_id','created_at']].sort_values('created_at'),
    all_claims[['user_id','created_at','claims_cum']],
    on='created_at', by='user_id', direction='backward', allow_exact_matches=False
).set_index('claim_id')

df['total_past_claims'] = df['claim_id'].map(m_c['claims_cum']).fillna(0).astype(int)
df['claim_rate'] = df['total_past_claims'] / df['total_past_orders'].replace(0, np.nan)
```

```python
history_s = history_full.sort_values('decision_at').copy()
history_s['cum_fraud'] = history_s.groupby('user_id')['is_fraud'].cumsum()
history_s['cum_known'] = history_s.groupby('user_id').cumcount() + 1

m_h = pd.merge_asof(
    df[['claim_id','user_id','created_at']].sort_values('created_at'),
    history_s[['user_id','decision_at','cum_fraud','cum_known']],
    left_on='created_at', right_on='decision_at', 
    by='user_id', direction='backward', allow_exact_matches=False
).set_index('claim_id')

df['past_fraud_claims'] = df['claim_id'].map(m_h['cum_fraud']).fillna(0)
df['past_known_claims'] = df['claim_id'].map(m_h['cum_known']).fillna(0)
df['past_fraud_rate'] = np.where(
    df['past_known_claims'] > 0,
    df['past_fraud_claims'] / df['past_known_claims'], 
    np.nan
)
```

```python
df['claim_amount_vs_avg_past'] = df['claim_amount'] / (df['total_order_amount'] / df['total_past_orders'].replace(0, np.nan))
df['order_amount_vs_avg_past'] = df['order_amount'] / df['avg_order_amount']
```

```python
df_train = df[df['is_test'] == 0].copy().sort_values('created_at').reset_index(drop=True)
df_test  = df[df['is_test'] == 1].copy()

q80 = df_train['created_at'].quantile(0.8)
train = df_train[df_train['created_at'] <= q80].copy()
val   = df_train[df_train['created_at'] > q80].copy()
```

```python
DROP = ['claim_id','user_id','order_id','created_at','order_created_at','order_delivered_at',
        'delivered_at','is_fraud','is_test','registered_at','decision_at','last_order_ts',
        'created_at_order', 'delivered_at_order']
FEATURES = [c for c in df.columns if c not in DROP]
FEATURES = [c for c in FEATURES if not pd.api.types.is_datetime64_any_dtype(df[c])]

cat_cols = df[FEATURES].select_dtypes(include=['object']).columns.tolist()

bool_cols = df[FEATURES].select_dtypes(include=['bool']).columns.tolist()
num_cols = [c for c in FEATURES if c not in cat_cols + bool_cols]

print('Категориальные:', cat_cols)
print('Числовые:', len(num_cols))
```

```python
num_pipe = Pipeline([('imp', SimpleImputer(strategy='median')),
                     ('sc',  StandardScaler())])
cat_pipe = Pipeline([('imp', SimpleImputer(strategy='constant', fill_value='missing')),
                     ('oh',  OneHotEncoder(handle_unknown='ignore', sparse_output=False))])
pre = ColumnTransformer([('num', num_pipe, num_cols),
                         ('cat', cat_pipe, cat_cols),
                         ('bool','passthrough', bool_cols)])

models = {}
models['LogReg'] = Pipeline([
    ('pre', pre),
    ('clf', LogisticRegression(max_iter=2000, class_weight='balanced', random_state=RANDOM_STATE))
]).fit(X_train, y_train)
models['RandomForest'] = Pipeline([
    ('pre', pre),
    ('clf', RandomForestClassifier(n_estimators=400, max_depth=12,
                                   class_weight='balanced_subsample',
                                   n_jobs=-1, random_state=RANDOM_STATE))
]).fit(X_train, y_train)
```

```python
Xtr_cb, Xval_cb = X_train.copy(), X_val.copy()
for c in num_cols:
    med = Xtr_cb[c].median()
    Xtr_cb[c]  = Xtr_cb[c].fillna(med)
    Xval_cb[c] = Xval_cb[c].fillna(med)
for c in cat_cols:
    Xtr_cb[c]  = Xtr_cb[c].astype(str)
    Xval_cb[c] = Xval_cb[c].astype(str)

models['CatBoost'] = CatBoostClassifier(
    iterations=800, depth=6, learning_rate=0.05,
    auto_class_weights='Balanced', random_seed=RANDOM_STATE,
    verbose=0, eval_metric='PRAUC'
).fit(Xtr_cb, y_train, cat_features=cat_cols,
      eval_set=(Xval_cb, y_val), use_best_model=True)

rows = []
proba_val = {}
for name, m in models.items():
    Xv = Xval_cb if name == 'CatBoost' else X_val
    p = m.predict_proba(Xv)[:, 1]
    proba_val[name] = p
    pred = (p >= 0.5).astype(int)
    rows.append({
        'model': name,
        'ROC-AUC': round(roc_auc_score(y_val, p), 4),
        'PR-AUC':  round(average_precision_score(y_val, p), 4),
        'F1':  round(f1_score(y_val, pred, zero_division=0), 4),
    })
print(pd.DataFrame(rows).sort_values('PR-AUC', ascending=False))
```

```python
def threshold_table(y_true, proba, thresholds, budget=0.15):
    n = len(y_true)
    out = []
    for t in thresholds:
        pred = (proba >= t).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, pred).ravel()
        out.append({
            'Threshold': round(t, 2),
            'Review %':  round((tp + fp) / n * 100, 2),
            'Precision': round(precision_score(y_true, pred, zero_division=0), 4),
            'Recall':    round(recall_score(y_true, pred, zero_division=0), 4),
            'F1':        round(f1_score(y_true, pred, zero_division=0), 4),
            'TP': int(tp), 'FP': int(fp), 'FN': int(fn),
        })
    tab = pd.DataFrame(out)
    tab['ok_budget'] = tab['Review %'] <= budget * 100
    return tab

thr_grid = np.round(np.arange(0.05, 0.96, 0.05), 2)

tab = threshold_table(y_val, proba_val['CatBoost'], thr_grid)
display(tab)

ok = tab[tab['ok_budget']]

BEST_THR = float(ok.sort_values(['TP','Precision'], ascending=False).iloc[0]['Threshold'])
print(f'\nВыбранный threshold: {BEST_THR}')
```
